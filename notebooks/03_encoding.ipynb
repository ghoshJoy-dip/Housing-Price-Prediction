```python
import pandas as pd
from category_encoders import TargetEncoder

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
train = pd.read_csv('C:/Projects/Housing-Price-Prediction/data/processed/train_clean.csv')
val = pd.read_csv('C:/Projects/Housing-Price-Prediction/data/processed/val_clean.csv')
test = pd.read_csv('C:/Projects/Housing-Price-Prediction/data/processed/test_clean.csv')
```

```python
train['date'] = pd.to_datetime(train['date'])
val['date'] = pd.to_datetime(val['date'])
test['date'] = pd.to_datetime(test['date'])
```

```python
def add_date_features(df):
    df['year'] = df['date'].dt.year
    df['quarter'] = df['date'].dt.quarter
    df['month'] = df['date'].dt.month

    df.insert(1, 'year', df.pop('year'))
    df.insert(2, 'quarter', df.pop('quarter'))
    df.insert(3, 'month', df.pop('month'))

    return df
```

```python
train = add_date_features(train)
val = add_date_features(val)
test = add_date_features(test)
```

Traget encoding city column

only fitting on train data and just transforming val and test, so there is no data leakage

```python
te = TargetEncoder(cols=['city'])

train['city_encoded'] = te.fit_transform(train['city'], train['price'])
val['city_encoded'] = te.transform(val['city'])
test['city_encoded'] = te.transform(test['city'])
```

```python
print(train[['city', 'city_encoded']].head())
```

```python
cols = ['date', 'median_sale_price', 'city']
train.drop(columns = cols, inplace = True)
val.drop(columns = cols, inplace = True)
test.drop(columns = cols, inplace = True)
```
