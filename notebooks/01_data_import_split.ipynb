#### Train-Validation-Test split

Temporal splitting of data for preventing data leakage:

2012 - 2019 -> Train set

2020 - 2021 -> Validation set

2022 - 2023 -> Test set

```python
# 1. sort data by date
df = df.sort_values('date')
```

```python
# 2. define validation and training cutoff
val_cutoff = '2020-01-01'
test_cutoff = '2022-01-01'
```

```python
# Train split
train_df = df[df['date'] < val_cutoff]
# Val split
val_df = df[(df['date'] >= val_cutoff) & (df['date'] < test_cutoff)]
# Test split
test_df = df[df['date'] >= test_cutoff]

print('Train shape', train_df.shape)
print('Val shape', val_df.shape)
print('Test shape', test_df.shape)
```

```python
train_df.to_csv('C:/Projects/Housing-Price-Prediction/data/raw/train.csv', index = False)
```

```python
val_df.to_csv('C:/Projects/Housing-Price-Prediction/data/raw/val.csv', index = False)
```

```python
test_df.to_csv('C:/Projects/Housing-Price-Prediction/data/raw/test.csv', index = False)
```
